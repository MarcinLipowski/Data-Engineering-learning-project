#### z-score

```python
import sklearn.preprocessing as skp

matrix = np.random.randint(0, 101, (5, 4))

vec = matrix[:, -1:] *100

vec_static = np.array([1 for i in range (5)])
vec_static = vec_static.reshape(5, 1)

matrix = np.concatenate((matrix, vec), axis=1)
matrix = np.concatenate((matrix, vec_static), axis=1)

matrix_scaled = skp.StandardScaler().fit_transform(matrix)
print(f"{matrix}\n")

mean = np.mean(matrix, axis = 0) # axis 1 - wiersz | 0 - kolumna
print(f"mean:\n{mean}\n")

std = np.std(matrix, axis = 0)
mask = np.isclose(std, 0)
print(F"std:\n{std}\nmask:\n{mask}\n")

result = matrix - mean

if mask.any():
    std = np.where(mask, 1, std) # nowa tablica z 1 tam gdzie true, reszta z std
    print(f"std with mask:\n{std}\n")

result /= std

print(f"result:\n{result}\n\nmatrix_scaled:\n{matrix_scaled}\n\n")

if np.allclose(result, matrix_scaled):
    print("Correct")
else:
    print("Fail")
```

#### Matrix multiply

```python
A = np.random.randint(-99, 100, (250, 200))
B = np.random.randint(-99, 100, (200, 300))


def matmul_naive(A, B):
    if len(A[0]) == len(B):
        C = np.zeros((len(A), len(B[0])))

        for i in range(len(A)):
            for j in range(len(B.T)):
                temp = 0
                for k in range(len(A[0])):
                    temp += A[i, k] * B[k, j]
                C[i, j] = temp
        return C
    else:
        print("Invalid matrix size.")
        return False
print("matmul_naive: ")
%timeit -n 1 -r 3 matmul_naive(A, B)

print("\n A @ B: ")
%timeit A @ B

print("\n Af @ Bf: ")
Af, Bf = A.astype(float), B.astype(float)
%timeit Af @ Bf
```

#### Max element

```python
X = np.random.randint(-10, 11, (10, 10))
def max(X):
    return np.unravel_index(np.argmax(X), (X.shape))

print(f"{X}\n\n")
max(X)
```
